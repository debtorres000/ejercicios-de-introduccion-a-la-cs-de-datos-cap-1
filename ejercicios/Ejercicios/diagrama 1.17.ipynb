```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(8, 10))

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis("off")

def crear_caja(texto, x, y):
    caja = FancyBboxPatch(
        (x, y), 4, 1,
        boxstyle="round,pad=0.05",
        linewidth=2,
        facecolor="white"
    )
    ax.add_patch(caja)
    ax.text(
        x + 2, y + 0.5,
        texto,
        ha="center",
        va="center",
        fontsize=13
    )

crear_caja("Data Sources", 3, 8)
crear_caja("Extract", 3, 6.5)
crear_caja("Transform", 3, 5)
crear_caja("Load", 3, 3.5)
crear_caja("Data Warehouse", 3, 2)

flechas = [
    (5, 8, 5, 7.5),
    (5, 6.5, 5, 6),
    (5, 5, 5, 4.5),
    (5, 3.5, 5, 3)
]

for x1, y1, x2, y2 in flechas:
    ax.annotate(
        "",
        xy=(x2, y2),
        xytext=(x1, y1),
        arrowprops=dict(arrowstyle="->", linewidth=2)
    )

ax.text(
    5, 9.5,
    "Proceso ETL",
    ha="center",
    fontsize=18,
    fontweight="bold"
)

plt.show()
```
